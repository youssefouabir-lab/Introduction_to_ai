```
def multiply_numbers(z, b):
    return z * b

print(multiply_numbers(2, 8))
```

```
def divide_numbers(a, b):
    if b != 0:
        return a / b
    else:
        return 'Cannot divide by zero'

print(divide_numbers(10, 2))
```
