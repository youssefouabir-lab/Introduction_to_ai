Task 3: Handling Null Values

```python
# Create DataFrame with null values in each column
df_null = pd.DataFrame({
    'Name': ['Alice', np.nan, 'Charlie', 'David'],
    'Age': [25, 30, np.nan, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', np.nan]
})

print("Original DataFrame with nulls:")
print(df_null)

# Method 1: Fill numeric nulls with 0 and categorical nulls with 'Unknown'
df_filled = df_null.copy()
df_filled['Age'] = df_filled['Age'].fillna(0)
df_filled['Name'] = df_filled['Name'].fillna('Unknown')
df_filled['City'] = df_filled['City'].fillna('Unknown')

print("\nFilled DataFrame:")
print(df_filled)

# Method 2: Demonstrate dropping rows with null values
df_cleaned = df_null.dropna()

print("\nDataFrame with dropped null rows:")
print(df_cleaned)
```

```python
# DataFrames provided
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Vertical concatenation of df1 and df2
df_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

# Horizontal merge with df3
df_final = pd.concat([df_concat, df3], axis=1)

print("Concatenated and Merged DataFrame:")
print(df_final)
print("\nFinal Shape:", df_final.shape)
```

Task 9: Iterators and Generators

```python
# Custom Iterator Class for first 5 even numbers
class EvenIterator:
    def __init__(self, limit=5):
        self.limit = limit
        self.count = 0
        self.num = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.limit:
            current = self.num
            self.num += 2
            self.count += 1
            return current
        else:
            raise StopIteration

# Generator function for first 5 even numbers
def even_generator(limit=5):
    count = 0
    num = 0
    while count < limit:
        yield num
        num += 2
        count += 1

print("Iterator output:")
even_iter = EvenIterator()
for val in even_iter:
    print(val, end=" ")
print()

print("\nGenerator output:")
for val in even_generator():
    print(val, end=" ")
print()
```

Task 10: Map, Reduce, and Filter

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# 1. Map to square all numbers
squared_numbers = list(map(lambda x: x ** 2, numbers))

# 2. Reduce to find the product of all numbers
product_numbers = reduce(lambda x, y: x * y, numbers)

# 3. Filter to get only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original Numbers:", numbers)
print("Squared Numbers (map):", squared_numbers)
print("Product of Numbers (reduce):", product_numbers)
print("Even Numbers (filter):", even_numbers)
```

Task 11: Object-Oriented Programming - Creating a Class

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

# Instantiate two Rectangle objects
rect1 = Rectangle(10, 5)
rect2 = Rectangle(7, 3)

print(f"Rectangle 1 (Length: {rect1.length}, Width: {rect1.width}) -> Area: {rect1.area()}, Perimeter: {rect1.perimeter()}")
print(f"Rectangle 2 (Length: {rect2.length}, Width: {rect2.width}) -> Area: {rect2.area()}, Perimeter: {rect2.perimeter()}")
```

Task 12: Pandas Data Analysis

```python
# DataFrame provided
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# 1. Average salary by department
avg_salary_dept = df_employees.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
print(avg_salary_dept)

# 2. Names of employees with salary > 60000
high_earners = df_employees[df_employees['Salary'] > 60000]['Name'].tolist()
print("\nEmployees with salary > 60,000:")
print(high_earners)

# 3. Add 'Bonus' column (10% of salary)
df_employees['Bonus'] = df_employees['Salary'] * 0.10
print("\nUpdated DataFrame with Bonus:")
print(df_employees)
```
