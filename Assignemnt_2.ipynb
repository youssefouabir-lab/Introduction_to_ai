```python
from google.colab import userdata

# creds come from Colab Secrets, nothing typed in the notebook
token = userdata.get("GITHUB_TOKEN")
user = "youssefouabir-lab"
repo = "colab-git-assignment2-oy"

# tell git who I am
!git config --global user.email "youssef.ouabir@gmail.com"
!git config --global user.name "youssefouabir-lab"

#repo folder
%cd /content/{repo}
!ls

!if [ -f Assignemnt_2.ipynb ]; then git mv Assignemnt_2.ipynb Assignment_2.ipynb; fi

# stage + commit
!git add Assignment_2.ipynb
!git commit -m "Add notebook with multiplication function" || echo "nothing new to commit"

# make sure the branch is called main
!git branch -M main

# clean remote url, no token saved in .git/config
!git remote remove origin
!git remote add origin https://github.com/{user}/{repo}.git

# pull the README that GitHub already has, then push
!git pull --no-rebase --allow-unrelated-histories --no-edit https://{user}:{token}@github.com/{user}/{repo}.git main
!git push https://{user}:{token}@github.com/{user}/{repo}.git main
```
